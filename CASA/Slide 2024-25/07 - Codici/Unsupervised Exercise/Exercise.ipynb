````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans, DBSCAN
from sklearn.preprocessing import StandardScaler

# Step 1: Generate a more challenging toy dataset
X, _ = make_blobs(n_samples=400, centers=[(-5, -5, 3), (0, 0, -3), (5, 5, 2), (8, -3, -2)], cluster_std=[1.2, 1.5, 1.0, 1.8], random_state=42)
X = StandardScaler().fit_transform(X)  # Normalize features

# Step 2: Apply K-Means clustering
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans_labels = kmeans.fit_predict(X)

# Step 3: Apply DBSCAN clustering
dbscan = DBSCAN(eps=0.5, min_samples=5)
dbscan_labels = dbscan.fit_predict(X)

# Step 4: Visualize results
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].scatter(X[:, 0], X[:, 1], c=kmeans_labels, cmap='viridis', marker='o')
axes[0].set_title("K-Means Clustering")

axes[1].scatter(X[:, 0], X[:, 1], c=dbscan_labels, cmap='plasma', marker='o')
axes[1].set_title("DBSCAN Clustering")

plt.show()
````
